## 📦 Step 1 — Imports

```python
# Import every function we need from our utility module
from my_weather_tacker import (
    init_csv,
    load_observations,
    display_menu,
    record_observation,
    view_statistics,
    search_by_date,
    view_all_observations,
    display_temperature_trend,
    filter_by_month,
    filter_by_season,
    predict_tomorrow,
    compare_year_over_year,
    display_outliers,
)

print("✅ All imports successful!")
```

## 🗂️ Step 2 — App Startup
This cell initializes the CSV file and loads any existing observations.  
Run this **once** before launching the app.

```python
# Initialize CSV (creates it if it doesn't exist) and load data
init_csv()
observations = load_observations()

print(f"\n📊 Observations loaded: {len(observations)}")
```

---
## 🚀 Step 3 — Run the App
This is the main application loop.  
It shows the menu and routes the user to the right function until they choose to exit.

> **For the demo:** Run this cell, then interact with the prompts below the cell.

```python
# Reload observations fresh before starting the loop
observations = load_observations()

# Main application loop
while True:
    choice = display_menu()

    if choice == "1":
        # Record a new observation and reload data so stats stay current
        record_observation()
        observations = load_observations()

    elif choice == "2":
        view_statistics(observations)

    elif choice == "3":
        search_by_date(observations)

    elif choice == "4":
        view_all_observations(observations)

    elif choice == "5":
        print("\n👋 Goodbye! Stay weather-aware.\n")
        break

    else:
        print("\n⚠️  Invalid choice. Please enter a number between 1 and 5.")
```
