## Stretch Goals

### Stretch 1 — Display Temperature Trends

Prints an ASCII dot chart of temperature over time — one row per observation.

```python
wt.displayTrends()
```

### Stretch 2 — Filter by Month

Returns only observations from a given month number (1–12) and shows their statistics.

```python
april_obs = wt.filter_by_month(observations, 4)  # change 4 to any month number
print(f"Found {len(april_obs)} observation(s) for April.")
wt.view_statistics(april_obs)
```

### Stretch 3 — Filter by Season

Returns only observations from a given season (Winter, Spring, Summer, Fall) and shows their statistics.

```python
spring_obs = wt.filter_by_season(observations, "Spring")  # change to Winter/Summer/Fall
print(f"Found {len(spring_obs)} Spring observation(s).")
wt.view_statistics(spring_obs)
```

### Stretch 4 — Predict Tomorrow's Weather

Uses **LinearRegression** to predict temperature and **KNeighborsClassifier** to predict the condition, both trained on month number from historical data.

```python
wt.predict_tomorrow(observations)
```

### Stretch 5 — Compare Years

Groups observations by year and prints avg/min/max temperature for each year, marking the current year.

```python
wt.compare_years(observations)
```

### Stretch 6 — Check Record-Breaking Temperatures

Finds the single hottest observation across all data. If it exceeds 20°C, it is flagged as a record breaker.

```python
wt.check_record()
```
