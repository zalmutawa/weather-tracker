# Weather Tracker — Demo Notebook

This notebook demonstrates all functions defined in `my_weather_tacker.py`.

| # | Function | Description |
|---|----------|-------------|
| 1 | `init_csv()` | Create or load the observations CSV file |
| 2 | `load_observations()` | Read all rows from the CSV |
| 3 | `view_statistics(obs)` | Summary stats (avg/min/max temp, top condition) |
| 4 | `filter_by_month(obs, month)` | Filter observations by month number |
| 5 | `filter_by_season(obs, season)` | Filter observations by season name |
| 6 | `predict_tomorrow(obs)` | ML-based temperature & condition prediction |
| 7 | `search()` | Search by date *(interactive — requires input)* |
| 8 | `displayTrends()` | ASCII temperature chart *(interactive)* |
| 9 | `recordObservation()` | Add a new observation *(interactive — requires input)* |

```python
import importlib
import my_weather_tacker as wt
importlib.reload(wt)  # ensures the latest version of the module is always loaded
```

## 1. Initialize the CSV File

`init_csv()` checks whether `observations.csv` exists. If not, it creates it with the correct header row.

```python
wt.init_csv()
```

## 2. Load Observations

`load_observations()` reads every row from the CSV and returns them as a list of dicts with numeric fields already cast to `float`.

```python
observations = wt.load_observations()
print(f"Loaded {len(observations)} observation(s)")
observations
```

## 3. View Statistics

`view_statistics(observations)` prints total count, average/min/max temperature, and the most common weather condition.

```python
wt.view_statistics(observations)
```

## 4. Filter by Month

`filter_by_month(observations, month)` takes an integer month number (1–12) and returns only the matching rows.

```python
april_obs = wt.filter_by_month(observations, 4)  # 4 = April
print(f"April observations: {len(april_obs)}")
wt.view_statistics(april_obs)
```

## 5. Filter by Season

`filter_by_season(observations, season)` returns only the rows that belong to the given season (`"Winter"`, `"Spring"`, `"Summer"`, or `"Fall"`).

```python
spring_obs = wt.filter_by_season(observations, "Spring")
print(f"Spring observations: {len(spring_obs)}")
wt.view_statistics(spring_obs)
```

## 6. Predict Tomorrow's Weather

`predict_tomorrow(observations)` trains a **LinearRegression** model to predict temperature and a **KNeighborsClassifier** to predict the weather condition, both based on month.

```python
# predict_tomorrow() uses LinearRegression for temperature and KNN for condition
wt.predict_tomorrow(observations)
```

## 7. Interactive Functions

The functions below require user input via `input()`.  
Run them in the terminal or execute the cell and type into the prompt that appears.

```python
# search() — prompts for a date (MM-DD-YYYY) and prints matching rows
wt.search()
```

```python
# displayTrends() — reads weather_data.csv and prints an ASCII dot chart of temperatures by date
wt.displayTrends()
```

```python
# recordObservation() — prompts for date, temp, condition, humidity, wind speed and saves to CSV
wt.recordObservation()
```
