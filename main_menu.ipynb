```python
import importlib
import my_weather_tacker as wt
importlib.reload(wt) #This is added to make sure that the kernel is not cached and we get the latest update on the script
wt.init_csv()
```

```python
while True:
    print("\n=== Weather Tracker ===")
    print("\n--- Zahraa's Components ---")
    print("1. Record a new observation")
    print("2. Search observations by date")
    print("3. Display temperature trends")

    print("\n--- Hadi's Components ---")
    print("4. View all observations")
    print("5. Check record-breaking temperatures")

    print("\n--- Zainab's Components ---")    
    print("6. View weather statistics")
    print("7. Filter by month")
    print("8. Filter by season")
    print("9. Predict tomorrow's weather")
    print("10. Compare current year with previous years")
    
    print("11. Exit")

    choice = input("Enter your choice (1-11): ")

    if choice == '1':
        wt.recordObservation()
    elif choice == '2':
        wt.search()
    elif choice == '3':
        wt.displayTrends()
    elif choice == '4':
        wt.view_all_observations()
    elif choice == '5':
        wt.check_record()
    elif choice == '6':
        observations = wt.load_observations()
        wt.view_statistics(observations)
    elif choice == '7':
        observations = wt.load_observations()
        try:
            month = int(input("Enter month number (1-12): "))
            results = wt.filter_by_month(observations, month)
            print(f"\nFound {len(results)} observation(s).")
        except ValueError:
            print("Invalid month. Please enter a number between 1 and 12.")
    elif choice == '8':
        observations = wt.load_observations()
        season = input("Enter season (Winter, Spring, Summer, Fall): ")
        results = wt.filter_by_season(observations, season)
        print(f"\nFound {len(results)} observations.")
    elif choice == '9':
        observations = wt.load_observations()
        wt.predict_tomorrow(observations)
    elif choice == '11':
        print("Thank you for using Weather Tracker. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 11.")
```
