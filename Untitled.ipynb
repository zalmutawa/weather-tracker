```python
def display_menu():
    """Display the main menu options."""
    print("\n=== Weather Tracker ===")
    print("1. Add a new Weather stats")
    print("2. View weather stats")
    print("3. Search for weather stats by date")
    print("4. view all observations")
    print("5. Exit")
    return input("Enter your choice (1-5): ")
```

```python
def main():
    """Main application function."""
    print("Welcome to Weather Tracker!")
    print("This app helps track the weather.")

    while True:
        choice = display_menu()

        if choice == '1':
            add_weather()
        elif choice == '2':
           view_weather()
        elif choice == '3':
            search_weather()
        elif choice == '4':
            view_observation()
        elif choice == '5':
            print("Thank you for using Weather Tracker. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 5.")
    
```
