```python
import csv
from tabulate import tabulate

def load_observations():
    """
    Read all rows from the CSV file and return them as a list of dicts.
    """
    observations = []
    with open('observations.csv', mode="r", newline="\n") as f:
        reader = csv.DictReader(f)
        for row in reader:
            # Cast numeric columns so math works later
            row["temperature_c"]   = float(row["temperature_c"])
            row["humidity_pct"]    = float(row["humidity_pct"])
            row["wind_speed_kmh"]  = float(row["wind_speed_kmh"])
            observations.append(row)
    if observations:
        print(tabulate(observations, headers="keys", tablefmt="fancy_grid"))
    else:
        print("No data available.")

    return observations
```

```python
def check_record():

    obs = load_observations()
    world_record = 34
    world_record_obs={}
    for obs in obs:
        if obs["temperature_c"]> world_record:
            world_record_obs = {'date':obs['date'] , 'temp':obs['temperature_c'] , 'conditon':obs['condition'] , 'humidity':obs['humidity_pct'], 'wind speed':obs['wind_speed_kmh']}
            
            world_record = obs["temperature_c"]
    if world_record_obs:
        print(f'NEW RECORD OF {world_record} DEGREES')
        return world_record_obs
    else:
        print('No tempreture broke the world record')
        
    
            

       
```
